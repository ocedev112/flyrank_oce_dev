## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Decision my work improves:**\
Which content items a reviewer looks at first, and with what mental model. Clustering doesn't score or rank individual pages it groups them into archetypes so a reviewer knows what kind of page they're dealing with before deciding an action, instead of triaging 500K+ pages one at a time from scratch.

**Who acts on it?**\
A content strategist or SEO reviewer with limited time/capacity, deciding which segment of the content inventory to prioritize this cycle. Secondary: whoever allocates team resourcing, since knowing "we have 4,000 stale visible pages vs. 200 hidden gems" shapes staffing/priorities, not just individual edits.

**Wrong Call**\
**Mislabeled archetype**:a real opportunity gets ignored silent, hard-to-detect cost.

**A healthy page flagged into a "needs rewrite" cluster**:wasted reviewer time on the wrong page.

**A page wrongly grouped into a prune/merge archetype**: if acted on, risks losing traffic/rankings that were actually fine.the costliest failure, since it's an irreversible action taken on trust in the cluster label.

**Unstable clusters**: every downstream action built on them is unreliable, regardless of how confident any single label looks.

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

**What this work will be able to say:**

**Observed:** These content items share statistically similar patterns across [impressions, CTR, engagement, position, freshness, etc.]." The clusters describe what happened in the datareal, measurable co-occurrence of signals not why it happened.

**Directional:** "Pages in this cluster tend to have low visibility despite strong engagement when they are seen" or "this archetype trends toward decay after a long period without updates." These are patterns observed across many pages, not guarantees about any one page.

**Decision-support:** "This archetype is a reasonable starting point for a reviewer to prioritize triagefor example, treating 'hidden gem' pages as promotion candidates and 'stale visible' pages as refresh candidates." The clusters narrow down where a human should look first; they don't replace the human's judgment on any individual page.

**Stability-qualified:** "These groupings replicate across re-runs / resampling," if that's true which supports treating the archetypes as a real structure in the data rather than an artifact of one clustering run.

**What this work will never be able to say:**

**Causal proof.** Never "refreshing pages in this cluster will cause recovery" or "this pattern causes traffic decline." Clustering shows association, not causation proving causation would require an actual experiment, which this data can't provide.\
**Predicting Google's algorithm.** Never "this is a ranking factor" or "this is how Google evaluates content." The data reflects observed outcomes it says nothing about why Google's systems produced those outcomes.\
**AI citation/ranking claims.** Never "this content is or isn't cited by AI systems" the only AI-related signal here is AI-referral sessions, not visibility inside AI answers themselves.\
**Semantic clustering.** Never call this "semantic" clustering there's no article text in the release, so groupings are based on metrics and metadata, not meaning. Calling it semantic would claim something the method didn't do.
Guaranteed outcomes for one page. An archetype describes a group tendency, not a guarantee for any single page in that group"this page is a hidden gem" is a probabilistic label, not a certainty.
